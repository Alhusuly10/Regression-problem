```python
import requests
import csv
import os

# Replace with your Solana RPC provider
SOLANA_RPC_URL = "https://mainnet.helius-rpc.com/?api-key=b07e3c73-a86e-404b-8b50-9393acaa38d3"
JUPITER_PROGRAM_ID = "JUP6LkbZbjS1jKKwapdHNy74zcZ3tLUZoi5QNyVTaV4"  # Jupiter Swap Program ID

# Function to fetch swap transactions
def fetch_swap_transactions():
    params = {
        "jsonrpc": "2.0",
        "id": 1,
        "method": "getSignaturesForAddress",
        "params": [JUPITER_PROGRAM_ID, {"limit": 1000}]
    }
    response = requests.post(SOLANA_RPC_URL, json=params).json()
    return response.get("result", [])

# Function to get wallet addresses from transaction details
def get_wallet_addresses():
    transactions = fetch_swap_transactions()
    wallet_addresses = set()

    for tx in transactions:
        signature = tx.get("signature")  # Get transaction hash
        if not signature:
            continue

        # Get transaction details to extract wallet address
        tx_details = requests.post(SOLANA_RPC_URL, json={
            "jsonrpc": "2.0",
            "id": 1,
            "method": "getTransaction",
            "params": [signature, "jsonParsed"]
        }).json()

        # Extract account addresses from the transaction
        if "result" in tx_details and tx_details["result"] is not None:
            accounts = tx_details["result"]["transaction"]["message"]["accountKeys"]
            if accounts:
                wallet_addresses.add(accounts[0])  # First address is usually the sender

    return list(wallet_addresses)

# Save wallet addresses to CSV
def save_to_csv(wallets, filename="eligible_wallets.csv"):
    with open(filename, mode="w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Wallet Address"])
        for wallet in wallets:
            writer.writerow([wallet])
    print(f"Eligible wallets saved to {filename}")
    os.startfile(filename)  # Open CSV automatically (Windows)

# Run the script
if __name__ == "__main__":
    eligible_wallets = get_wallet_addresses()
    save_to_csv(eligible_wallets)
```
